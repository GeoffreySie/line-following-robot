```python
import cv2

INPUT_AVI    = 'data/driving_data_2.avi'
OUTPUT_AVI   = 'data/driving_data_2_cropped.avi'
CROP_SECONDS = 4

cap = cv2.VideoCapture(INPUT_AVI)
fps    = cap.get(cv2.CAP_PROP_FPS)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

frames_to_skip = int(fps * CROP_SECONDS)
cap.set(cv2.CAP_PROP_POS_FRAMES, frames_to_skip)

fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter(OUTPUT_AVI, fourcc, fps, (width, height))

written = 0
while True:
    ret, frame = cap.read()
    if not ret:
        break
    out.write(frame)
    written += 1

cap.release()
out.release()
print(f'Skipped {frames_to_skip} frames ({CROP_SECONDS}s). Wrote {written} frames to "{OUTPUT_AVI}"')
```
